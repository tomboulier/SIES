```python
import numpy as np
import matplotlib.pyplot as plt
from sies import C2Boundary, MConfig, ElectricFish
theta = np.linspace(0, 2*np.pi, 16, endpoint=False)
src_rcv = np.vstack([np.cos(theta), np.sin(theta)]) * 2.0
obj_theta = np.linspace(0, 2*np.pi, 50, endpoint=False)
obj = C2Boundary(np.vstack([0.5*np.cos(obj_theta), 0.5*np.sin(obj_theta)]))
cfg = MConfig(sources=src_rcv, receivers=src_rcv)
cfg.Omega0 = C2Boundary(src_rcv)
fish = ElectricFish(obj, [1.0], [1.0], cfg)
plt.figure();
plt.plot(obj.points[0], obj.points[1], label='objet');
plt.scatter(src_rcv[0], src_rcv[1], c='r', label='capteurs');
plt.axis('equal');
plt.legend();
```

```python
def potential(x, source):
    return -np.log(np.linalg.norm(x - source[:, None], axis=0)) / (2*np.pi)
grid_x, grid_y = np.mgrid[-1:1:100j, -1:1:100j]
phi = potential(np.vstack([grid_x.ravel(), grid_y.ravel()]), src_rcv[:,0]).reshape(grid_x.shape)
plt.figure();
plt.pcolormesh(grid_x, grid_y, phi, shading='auto');
plt.plot(obj.points[0], obj.points[1], 'k');
plt.axis('equal');
```
